# Synthetic-only benchmark quick start

This notebook runs the public smoke path on generated ridge patterns. It downloads no biometric data and writes only to `.smoke-notebook/`.

```python
from pathlib import Path
import subprocess
import sys

repo = Path.cwd()
if not (repo / 'scripts' / 'smoke_synthetic.py').is_file():
    raise RuntimeError('Run this notebook with the repository root as the working directory')
subprocess.run([sys.executable, 'scripts/smoke_synthetic.py', '--work-dir', '.smoke-notebook'], check=True)
```
